```python
# Save the daily top 5 recommendations to CSV and validate older entries after 2 weeks
from datetime import datetime

if 'output_dir' not in globals():
    output_dir = os.path.abspath(os.getenv('OUTPUT_DIR', os.getcwd()))

if not os.path.isdir(output_dir):
    output_dir = os.path.abspath(os.getcwd())

recommendation_dir = os.path.join(output_dir, 'recommendations')
os.makedirs(recommendation_dir, exist_ok=True)
csv_path = os.path.join(recommendation_dir, "daily_recommendations.csv")
DATE_FORMAT = "%d%m%Y"
run_timestamp = datetime.now().strftime(DATE_FORMAT)
current_date = pd.Timestamp.today().normalize()

# Build a recommendation dataframe even if the earlier ranking variable was not preserved
if 'top10' not in globals() or top10 is None:
    top10 = pd.DataFrame()

if not isinstance(top10, pd.DataFrame):
    try:
        top10 = pd.DataFrame(top10)
    except Exception:
        top10 = pd.DataFrame()

if top10.empty and 'results' in globals() and results:
    ranked = pd.DataFrame(results).copy()
    if 'FinalScore' in ranked.columns:
        ranked = ranked.sort_values(by='FinalScore', ascending=False).reset_index(drop=True)
    top10 = ranked.head(15).copy()


def normalize_date_value(value):
    if isinstance(value, (datetime, pd.Timestamp)):
        return value.strftime(DATE_FORMAT)

    text = str(value).strip()
    if not text:
        return pd.NA

    try:
        parsed = pd.to_datetime(text, format=DATE_FORMAT, errors='coerce')
        if pd.notna(parsed):
            return parsed.strftime(DATE_FORMAT)
    except Exception:
        pass

    try:
        parsed = pd.to_datetime(text, errors='coerce')
        if pd.notna(parsed):
            return parsed.strftime(DATE_FORMAT)
    except Exception:
        pass

    return text


def validate_prior_recommendations(csv_path):
    if not os.path.exists(csv_path):
        return None

    prior_df = pd.read_csv(csv_path, dtype=str)
    if prior_df.empty:
        return prior_df

    prior_df = ensure_validation_columns(prior_df)
    prior_df['Date'] = prior_df['Date'].apply(normalize_date_value)
    prior_df['_Date_dt'] = pd.to_datetime(prior_df['Date'], format=DATE_FORMAT, errors='coerce')
    prior_df['Price'] = pd.to_numeric(prior_df['Price'], errors='coerce')

    for idx, row in prior_df.iterrows():
        if pd.isna(row['_Date_dt']) or pd.isna(row['Price']):
            continue

        age_days = (current_date - row['_Date_dt']).days
        prior_df.at[idx, 'Validation_Age_Days'] = int(age_days)

        if age_days >= 14 and pd.isna(row['Validation_Status']):
            symbol = f"{row['Stock']}.NS"
            try:
                hist = fetch_data(symbol)
                latest_price = hist['Close'].iloc[-1]
                change_pct = ((latest_price - row['Price']) / row['Price']) * 100 if row['Price'] > 0 else 0.0
                status = 'UP' if change_pct > 0.0 else ('DOWN' if change_pct < 0.0 else 'FLAT')
                prior_df.at[idx, 'Validation_Status'] = status
                prior_df.at[idx, 'Validation_Return_Pct'] = round(change_pct, 2)
                prior_df.at[idx, 'Validation_Checked_On'] = current_date.strftime(DATE_FORMAT)
            except Exception as e:
                prior_df.at[idx, 'Validation_Status'] = 'ERROR'
                prior_df.at[idx, 'Validation_Checked_On'] = current_date.strftime(DATE_FORMAT)

    return prior_df


if top10.empty:
    print("\nNo recommended stocks found to save.")
else:
    top5 = top10.head(5).copy()
    top5.insert(0, 'Date', run_timestamp)
    top5 = top5[[
        'Date', 'Stock', 'FinalScore', 'BaseScore', 'FactorScore', 'News_Score', 'AI_Score',
        'AI_Bias', 'AI_Confidence', 'Price', 'RSI', 'ADX', 'MACD_Signal', 'Volatility',
        'Liquidity', 'Alpha', 'Volume_Ratio', 'Breakout', 'Trend_Structure', 'Stop_Loss',
        'Selling_Price', 'Daily_Change', 'Daily_Change_Pct', 'High_Volume',
        'Extreme_RSI', 'Bearish_Div'
    ]]

    top5 = ensure_validation_columns(top5)
    top5['Validation_Status'] = 'PENDING'
    top5['Validation_Return_Pct'] = pd.NA
    top5['Validation_Checked_On'] = pd.NA
    top5['Validation_Age_Days'] = 0

    prior_df = validate_prior_recommendations(csv_path)
    if prior_df is None or prior_df.empty:
        combined_df = top5
    else:
        combined_df = pd.concat([prior_df, top5], ignore_index=True)
        combined_df = ensure_validation_columns(combined_df)
        combined_df['Date'] = combined_df['Date'].apply(normalize_date_value)
        combined_df['_Date_dt'] = pd.to_datetime(combined_df['Date'], format=DATE_FORMAT, errors='coerce')

    combined_df.to_csv(csv_path, mode='w', index=False)
    print(f"\nSaved updated daily recommendations with validation status to {csv_path}")

print("\n✅ ANALYSIS COMPLETE - READY FOR 1-2 WEEK SWING TRADES")
```
